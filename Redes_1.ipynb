Networkx incluye muchas funciones de generacion de graficos

```python
import networkx as nx
```

Graph crea un gráfico vacío sin nodos ni arcos.

```python
G=nx.DiGraph() # Grafica dirigida
```

El gráfico G se puede empezar de varias maneras. NetworkX incluye muchas funciones de generador de gráficos y facilidades para leer y escribir gráficos en muchos formatos.

```python
G
```

Para crea una conexion entre dos nodos, por ejemplo:


*   A -> B



Donde la arista representa una relación dirigida con un atributo llamado capacity

```python
G.add_edge("A","B",capacity=2)  # Conexión de A hacia B
```

La siguiente función nos permite dibujar en pantalla el estado actual del grafo.

```python
nx.draw_networkx(G) # Contiene los nodos A y B conectados por una arista dirigida
```

Añadiendo más conexiones entre los nodos:


*   A -> C
*   B -> D

*   B -> C
*   C -> D





```python
G.add_edge("A","C", capacity=2) # Conexión de A hacia C
G.add_edge("B","D",capacity=2) # Conexión de B hacia D
G.add_edge("B","C",capacity=2) # Conexión de B hacia C
G.add_edge("C","D",capacity=2) # Conexión de C hacia D
```

Estado actual del grafo:

```python
nx.draw_networkx(G) #Esta funcion devuelve la lista de nodos que conforman la red
```
