```python
# Condición inicial
x = 1 # Se encuentra en el sector naranja
y = 0 # Sector azul
z = 0 # Sector verde
t = 0 # Sector rojo

n = 25
```

```python
x_i = []
y_i = []
z_i = []
t_i = []

for i in range(n):
    xi = (1/4)*x + 0*y + (1/4)*z + (2/4)*t
    yi = (2/4)*x + (1/4)*y + 0*z + (1/4)*t
    zi = (1/4)*x + (2/4)*y + (1/4)*z + 0*t
    ti = 0*x + (1/4)*y + (2/4)*z + (1/4)*t

    x_i.append(xi)
    y_i.append(yi)
    z_i.append(zi)
    t_i.append(ti)

    x = xi
    y = yi
    z = zi
    t = ti
```
